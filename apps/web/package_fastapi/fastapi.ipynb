# fastapi

- starlette 웹 프레임워크 기반의 REST API 서버

```python
%%capture
%%writefile fastapi1/main.py

from fastapi import FastAPI

app = FastAPI()


@app.get("/")
async def root():
    return {"message": "Hello World"}
```

- 개발시에는 `fastapi dev 파일이름`
- 가동시에는 `fastapi run 파일이름`
- 파일이름이 main.py 인 경우에는 파일이름 생략 가능
