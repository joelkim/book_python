# fastapi

- starlette 웹 프레임워크 기반의 REST API 서버

```python
%%capture
%%writefile fastapi1.py

from fastapi import FastAPI

app = FastAPI()


@app.get("/")
async def read_root():
    return {"Hello": "World"}


@app.get("/items/{item_id}")
async def read_item(item_id: int, q: str | None = None):
    return {"item_id": item_id, "q": q}
```

- 개발시에는 `fastapi dev 파일이름`
- 가동시에는 `fastapi run 파일이름`
- 파일이름이 main.py 인 경우에는 파일이름 생략 가능
